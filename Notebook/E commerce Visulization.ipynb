# E-Commerce Customer Analysis — Visualizations

```python
import pandas as pd
import matplotlib.pyplot as plt

df_clean = pd.read_csv("../data/clean/customer_shopping_behavior_cleaned.csv")

df_clean.head()
```

## Customer Demographics

```python
plt.figure(figsize=(8, 5))

df_clean["Gender"].value_counts().plot(kind="bar")

for i, value in enumerate(df_clean["Gender"].value_counts()):
    plt.text(i,value,str(value),ha="center",va="bottom")

plt.title("Customer Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("../visualizations/gender_distribution.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

## Product & Category Analysis

```python
plt.figure(figsize=(8, 5))

category_sales = df_clean.groupby("Category")["Purchase Amount (USD)"].sum().sort_values(ascending=False)

category_sales.plot(kind="bar")

for i, value in enumerate(category_sales):
    plt.text(i, value, f"${value:,.0f}", ha="center", va="bottom")

plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales (USD)")

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("../visualizations/category_sales.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

top_products = (
    df_clean.groupby("Item Purchased")["Purchase Amount (USD)"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products.plot(kind="bar")

for i, value in enumerate(top_products):
    plt.text(i, value, f"${value:,.0f}", ha="center", va="bottom")

plt.title("Top 10 Products by Total Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales (USD)")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig("../visualizations/top_10_products_sales.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

## Seasonal Analysis

```python
plt.figure(figsize=(8, 5))

season_sales = (
    df_clean.groupby("Season")["Purchase Amount (USD)"]
    .sum()
    .sort_values(ascending=False)
)

season_sales.plot(kind="bar")

for i, value in enumerate(season_sales):
    plt.text(i, value, f"${value:,.0f}", ha="center", va="bottom")

plt.title("Total Sales by Season")
plt.xlabel("Season")
plt.ylabel("Total Sales (USD)")

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("../visualizations/season_sales.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

## Customer Satisfaction

```python
plt.figure(figsize=(8, 5))

rating_counts = df_clean["Review Rating"].value_counts().sort_index()

rating_counts.plot(kind="bar")

for i, value in enumerate(rating_counts):
    plt.text(i, value, str(value), ha="center", va="bottom")

plt.title("Customer Review Rating Distribution")
plt.xlabel("Review Rating")
plt.ylabel("Number of Reviews")

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("../visualizations/review_rating_distribution.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

## Purchase Behavior

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df_clean["Purchase Amount (USD)"],
    bins=15,
    edgecolor="black"
)

plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Number of Purchases")

plt.tight_layout()

plt.savefig(
    "../visualizations/purchase_amount_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
